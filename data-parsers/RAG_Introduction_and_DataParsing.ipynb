```python
from typing import Dict

# --- Example JSON Data for demonstration (assuming it's in /content/sample_data.json) ---
# If your sample_data.json is different, adjust these examples accordingly.
# For this demonstration, I'm assuming it contains an array of objects like:
# [
#   {
#     "id": 1,
#     "title": "LangChain Basics",
#     "content": "LangChain is a framework for developing applications powered by language models.",
#     "author": "AI Assistant"
#   },
#   {
#     "id": 2,
#     "title": "Advanced LangChain",
#     "content": "Dive deeper into agents, chains, and custom components.",
#     "author": "AI Assistant"
#   }
# ]

# 1. Using `content_key`
print("\n--- Using content_key ---")
json_loader_content_key = JSONLoader(
    file_path='/content/sample_data.json',
    content_key='content',
    jq_schema=".[]"
)
docs_content_key = json_loader_content_key.load()
print(f"Page Content (first doc): {docs_content_key[0].page_content}")
print(f"Metadata (first doc): {docs_content_key[0].metadata}")
```

```python
# 2. Using `metadata_func`
print("\n--- Using metadata_func ---")
def metadata_extractor(record: Dict, metadata: Dict) -> Dict:
    # Here you can define custom logic to extract and structure your metadata
    metadata['document_id'] = record.get('id')
    metadata['doc_title'] = record.get('title')
    metadata['source_author'] = record.get('author') # Custom key name
    return metadata

json_loader_metadata_func = JSONLoader(
    file_path='/content/sample_data.json',
    content_key='content', # Still need to specify content_key for the main text
    metadata_func=metadata_extractor
)
docs_metadata_func = json_loader_metadata_func.load()
print(f"Page Content (first doc): {docs_metadata_func[0].page_content}")
print(f"Metadata (first doc): {docs_metadata_func[0].metadata}")
```

```python
# 3. No `jq_schema` or `content_key` (Default Behavior)
print("\n--- Default behavior (no content_key or jq_schema) ---")
# This will stringify the entire JSON object as page content.
json_loader_default = JSONLoader(
    file_path='/content/sample_data.json',
)
docs_default = json_loader_default.load()
print(f"Page Content (first doc): {docs_default[0].page_content}")
print(f"Metadata (first doc): {docs_default[0].metadata}")
```
